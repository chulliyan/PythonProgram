```python
num1 = int(input("Enter num1 :"))
num2 = int(input("Enter num2 :"))
def larger(a , b):
  if a > b:
    print(f"{a} is larger than {b}")
  else:
    print(f"{b} is larger than {a}")

larger(num1, num2)




```
